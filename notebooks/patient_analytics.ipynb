```python
# Since each patients is visiting multiple times to have a unique identifier we need to indexes permanent
# Turn the index into a unique string (e.g., 'V0', 'V1', 'V2')
df['visit_id'] = 'V' + df.index.astype(str)
df.head()
```

#### Creating another table for secondary diagnoses as a patient can have multiple secondary diagnoses separated by '|'


```python
# 1. Creating Secondary diagnoses dataframe and dropping rows without secondary diseases
df_secondary_diag = df[['visit_id', 'patient_id', 'secondary_diagnoses', 'secondary_icd10s']].dropna()

# 2.  Separating '|' dilimited diseases into at list
df_secondary_diag['secondary_diagnoses'] = df_secondary_diag['secondary_diagnoses'].str.split('|') 
df_secondary_diag['secondary_icd10s'] = df_secondary_diag['secondary_icd10s'].str.split('|')

# 3. Creating the bridging table and mapping '|' separated diseases and codes to each other using explode
df_bridge_sec_diag = df_secondary_diag.explode(['secondary_diagnoses', 'secondary_icd10s'])

# 4. Renaming columns to 
df_bridge_sec_diag = df_bridge_sec_diag.rename(columns={'secondary_diagnoses': 'diagnosis_name', 
                                                       'secondary_icd10s': 'icd10_code'})

df_bridge_sec_diag.head(10)
```

```python
df_bridge_sec_diag.info()
```

```python
# Removing _ characters from text columns and making them in title case
df_bridge_sec_diag['diagnosis_name'] = df_bridge_sec_diag['diagnosis_name'].apply(lambda x: x.replace('_', ' ').title())
df_bridge_sec_diag.head()
```

```python
# Removing columns related to secondary diagnoses from primary diagnoses table
df_pri_diag = df.drop(columns=['secondary_diagnoses', 'secondary_icd10s'])
df_pri_diag.head()
```

```python
# Removing _ characters from text columns and making them in title case
columns = ['visit_type', 'primary_diagnosis', 'provider_specialty']
for column in columns:
    df_pri_diag[column] = df_pri_diag[column].apply(lambda x: x.replace('_', ' ').title())
df_pri_diag.head()
```

##### Note on Data Integrity: Handling Nulls in days_to_readmission
You will notice that the NaN values in the days_to_readmission column have been intentionally left as-is. In this dataset, a null value simply means the patient was never readmitted. If we were to impute these missing values with 0, the data would falsely imply that thousands of patients were readmitted on the exact same day they were discharged. This would severely skew our average readmission metrics downward. By preserving these structural nulls, downstream tools like SQL and Power BI will automatically ignore them when calculating averages, ensuring our final clinical metrics remain 100% accurate.

## Migrating Cleaned Data to MySQL

```python
# Importing Libraries
from sqlalchemy import create_engine
import pymysql
from urllib.parse import quote_plus

# Step 1: Connect to MySQL database
db_user = os.environ.get('DB_USER_PROJECT')
db_password = os.environ.get('DB_PASS_PROJECT')
db_host = os.environ.get('HOST')
db_port = os.environ.get('PORT_PROJECT')
database = 'patient_analytics_db'

# To parse special characters in password if any.
safe_password = quote_plus(db_password)

# Creating Engine
engine = create_engine(f"mysql+pymysql://{db_user}:{safe_password}@{db_host}:{db_port}/{database}")
print("Connected to MySQL successfully!")

# Step 2: Load tables into MySQL

# UPLOADING PARENT TABLES (WITHOUT FOREIGN KEYS)

# Charlson Comorbidity Index table
table_cci = 'charlson_comorbidity_index'
cci_df.to_sql(table_cci, engine, if_exists='append', index=False)
print(f"Data Successfully loaded into table '{table_cci}' in database '{database}'.")

# Patients table
table_patients = 'patients'
df_pat.to_sql(table_patients, engine, if_exists='append', index=False)
print(f"Data Successfully loaded into table '{table_patients}' in database '{database}'.")

# UPLOADING CHILD TABLES (WITH FOREIGN KEYS)

# Medicine table
table_meds = 'medicines'
meds_df.to_sql(table_meds, engine, if_exists='append', index=False)
print(f"Data Successfully loaded into table '{table_meds}' in database '{database}'.")

# Primary diagnoses table
table_primary_diag = 'diagnoses'
df_pri_diag.to_sql(table_primary_diag, engine, if_exists='append', index=False)
print(f"Data Successfully loaded into table '{table_primary_diag}' in database '{database}'.")

# Secondary diagnoses table
table_sec_diag = 'visit_secondary_diagnoses'
df_bridge_sec_diag.to_sql(table_sec_diag, engine, if_exists='append', index=False)
print(f"Data Successfully loaded into table '{table_sec_diag}' in database '{database}'.")

# Outcomes table
table_outcomes = 'outcomes'
outcomes_df.to_sql(table_outcomes, engine, if_exists='append', index=False)
print(f"Data Successfully loaded into table '{table_outcomes}' in database '{database}'.")

# UPLOADING THE MASSIVE LAB RESULTS TABLE (In Chunks)
print("Pushing 2.2 million lab results to MySQL (this may take a minute)...")

table_lab = 'lab_results'
lab_df.to_sql(table_lab, engine, if_exists='append', index=False, chunksize=50000)
print(f"Data Successfully loaded into table '{table_lab}' in database '{database}'.")

print("Data Pipeline Complete! All tables successfully migrated.")
```
